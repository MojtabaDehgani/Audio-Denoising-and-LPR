```python
import cv2, numpy as np

def sharpness(im):                            # This function and this code are exactly the same code in Part2 and Part3 of the Realistic Case section, which also has the exact same name. (def sharpness(im))
    gx, gy = np.gradient(im.astype(float)/255)
    return (gx**2 + gy**2).var()

n1 = cv2.imread('noisy/n1.jpg', cv2.IMREAD_GRAYSCALE)
n2 = cv2.imread('noisy/n2.jpg', cv2.IMREAD_GRAYSCALE)

print("sharpness n1 =", sharpness(n1))
print("sharpness n2 =", sharpness(n2))
```
